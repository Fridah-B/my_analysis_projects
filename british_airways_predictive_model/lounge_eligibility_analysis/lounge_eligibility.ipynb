```python
file_path = r"E:\my_analysis_projects\british_airways_predictive_model\British Airways Summer Schedule Dataset - Forage Data Science Task 1.xlsx"
b_a_data = pd.read_excel(file_path)
```

```python
#copying the file so as not to change the original file
ba_data = b_a_data.copy()
```

```python
ba_data.drop_duplicates()
```

```python
grouped_data = ba_data.groupby(["TIME_OF_DAY", "ARRIVAL_REGION"])[['FIRST_CLASS_SEATS','BUSINESS_CLASS_SEATS', 'ECONOMY_SEATS', 'TIER1_ELIGIBLE_PAX','TIER2_ELIGIBLE_PAX','TIER3_ELIGIBLE_PAX'  ]].sum()
```

```python
grouped_data.head()
```

```python
# new column for the total number of seats
grouped_data['total_seats'] = grouped_data[['FIRST_CLASS_SEATS','BUSINESS_CLASS_SEATS', 'ECONOMY_SEATS']].sum(axis=1)
```

```python
# checking percentage of eligible customers per tier
def eligible_customers(tier, total_seats):
    percentage = round((tier/total_seats)*100, 2)
    
    return percentage
```

```python
#new columns for percent of eligible customers
grouped_data['TIER_1_PERCENT'] = eligible_customers(grouped_data['TIER1_ELIGIBLE_PAX'], grouped_data['total_seats'])
grouped_data['TIER_2_PERCENT'] = eligible_customers(grouped_data['TIER2_ELIGIBLE_PAX'], grouped_data['total_seats'])
grouped_data['TIER_3_PERCENT'] = eligible_customers(grouped_data['TIER3_ELIGIBLE_PAX'], grouped_data['total_seats'])
```

```python
lookup_table = grouped_data.reset_index()
```

```python
lookup_table = lookup_table[['TIME_OF_DAY', 'ARRIVAL_REGION', 'TIER_1_PERCENT', 'TIER_2_PERCENT', 'TIER_3_PERCENT']]
```
